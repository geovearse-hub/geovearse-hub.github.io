### Egypt Population Dashboard
This dashboard combines satellite imagery with population statistics.

```python
import folium
import pandas as pd
import plotly.express as px
from IPython.display import display, HTML

# Using the existing 'df' variable from previous cells

# 1. Create the Satellite Map
m = folium.Map(location=[26.8206, 30.8025], zoom_start=6,
               tiles='https://mt1.google.com/vt/lyrs=s&x={x}&y={y}&z={z}',
               attr='Google Imagery')

for index, row in df.iterrows():
    folium.CircleMarker(
        location=[row['Lat'], row['Lon']],
        radius=row['Population'] / 250000,
        popup=f"<b>{row['City']}</b><br>Population: {row['Population']:,}",
        color='yellow', fill=True, fill_color='orange', fill_opacity=0.7
    ).add_to(m)

# 2. Create the Bar Chart
fig = px.bar(df.sort_values('Population', ascending=False),
             x='City', y='Population',
             color='Population',
             title='Population Comparison',
             template='plotly_dark')

# Display them
display(HTML('<h3>Interactive Satellite Map</h3>'))
display(m)
display(fig)
```

```python
import folium
import pandas as pd
import plotly.express as px
from folium import plugins

# Sample data for Egyptian cities and their approximate population
data = {
    'City': ['Cairo', 'Alexandria', 'Giza', 'Shubra El Kheima', 'Port Said', 'Suez', 'Luxor', 'Mansoura', 'Tanta', 'Asyut'],
    'Lat': [30.0444, 31.2001, 30.0131, 30.1286, 31.2653, 29.9668, 25.6872, 31.0409, 30.7865, 27.1783],
    'Lon': [31.2357, 29.9187, 31.2089, 31.2422, 32.3019, 32.5498, 32.6396, 31.3785, 31.0004, 31.1859],
    'Population': [9540000, 5160000, 4870000, 1165000, 751000, 660000, 422000, 621000, 508000, 462000]
}

df = pd.DataFrame(data)
```

```python
# Create a Folium map centered on Egypt with Satellite Imagery
m = folium.Map(location=[26.8206, 30.8025], zoom_start=6, tiles='https://mt1.google.com/vt/lyrs=s&x={x}&y={y}&z={z}', attr='Google Imagery')

# Add circles for population
for index, row in df.iterrows():
    folium.CircleMarker(
        location=[row['Lat'], row['Lon']],
        radius=row['Population'] / 200000,  # Scale radius for visibility
        popup=f"{row['City']}: {row['Population']:,}",
        color='crimson',
        fill=True,
        fill_color='crimson',
        fill_opacity=0.6
    ).add_to(m)

display(m)
```

```python
# Create a complementary bar chart for the dashboard
fig = px.bar(df, x='City', y='Population', title='Population of Major Egyptian Cities',
             color='Population', color_continuous_scale='Reds')
fig.show()
```
